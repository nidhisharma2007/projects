```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Bar chart

plt.bar(df['Location'], df['Area'])
plt.xlabel("Locatiom")
plt.ylabel("Area")
plt.title("Bar Chart ")
plt.show()
```

```python
#Histogram

plt.hist(df['Price'])
plt.xlabel("House Price")
plt.ylabel("Frequency")
plt.title("Histogram of House Prices")
plt.show()
```

```python
#Pie chart
price_count = df.groupby('Garage')['Price'].count()

plt.pie(price_count, labels=price_count.index, autopct='%1.1f%%')
plt.title("House Count by Type")
plt.show()
  
```

```python
#Line Chart
DF=df.head()
plt.plot(DF['YearBuilt'], DF['Price'])
plt.xlabel("Year")
plt.ylabel("Price")
plt.title("Line Chart")
plt.show()
```

```python
#Scatter plot

plt.scatter(df['Area'], df['Price'])
plt.xlabel("Area")
plt.ylabel("Price")
plt.title("Scatter Plot (Area vs Price)")
plt.show()
```

```python
#Subplot

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.hist(df['Price'])
plt.title("Histogram")

plt.subplot(2, 2, 2)
plt.scatter(df['Area'], df['Price'])
plt.title("Scatter Plot")

plt.subplot(2, 2, 3)
plt.plot(df.index, df['Price'])
plt.title("Line Chart")

plt.subplot(2, 2, 4)
plt.bar(df.index[:10], df['Price'][:10])
plt.title("Bar Chart")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns

```

```python
#Relplot

sns.relplot(x='Area', y='Price', data=df)
```

```python
#Histplot

sns.histplot(df['Price'])
plt.title("Histplot of House Prices")
plt.show()
```

```python
#KDE plot

sns.kdeplot(df['Price'])
plt.title("KDE Plot of House Prices")
plt.show()
```

```python
#Boxplot

sns.boxplot(y='Price', data=df)
plt.title("Boxplot of House Prices")
plt.show()
```
